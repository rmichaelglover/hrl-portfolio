# Legal geometry and contextual subboard ledger
Attack maps include defended friendly squares. Legal move maps enforce king safety, blocking, castling, en passant and promotion through python-chess. A cropped window is a heuristic record, not a solved subgame.

```python
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / "relaxfish.py").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
import chess
import numpy as np
from relaxfish import *
```

```python
board = chess.Board()
print("Pawn attack map:")
print(movement_layer(board, chess.E2))
print("Pawn legal destination map:")
print(movement_layer(board, chess.E2, legal=True))
ledger = SubboardLedger()
key = ledger.register(board, (0, 0), 4, evaluation=0.0)
print(ledger.records[key])
ledger.save(root / "examples" / "subboard-ledger.json")
```
