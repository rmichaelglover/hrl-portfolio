# Primary and dual tactical role architecture
Primary rows identify individual pieces by square. Dual rows are roles, with one column per actual piece. These are conditional distributions, not a one-to-one assignment.

```python
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / "relaxfish.py").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
import chess
import numpy as np
from relaxfish import *
```

```python
board = chess.Board()
model = ChessRelaxationLabeler(board)
result = model.solve()
print("Primary:", model.P_primary.shape, "Dual:", model.P_dual.shape)
print("Converged:", result["converged"], "Residual:", result["residual"])
for role, row in zip(ROLES, model.P_dual):
    print(role, "->", model.pieces[int(np.argmax(row))])
```
