# RelaxFish: distributed HRL chess profiles
Architecture: Michael Emanuel Glover. Polished prototype, October 25, 2026.

Independent positions run in worker processes; each worker preserves the complete board interaction graph. Numerical convergence does not certify chess optimality.

```python
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / "relaxfish.py").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
import chess
import numpy as np
from relaxfish import *
```

```python
fens = [chess.STARTING_FEN, "7k/5Q2/6K1/8/8/8/8/8 w - - 0 1"]
profiles = distributed_profiles(fens, workers=1)
for fen, result in zip(fens, profiles):
    print(chess.Board(fen))
    print("iterations:", result["iterations"], "residual:", result["residual"])
    for piece, weights in zip(result["pieces"], result["primary"]):
        best = int(np.argmax(weights))
        print(piece, ROLES[best], round(weights[best], 3))
```

For multiprocessing, call `distributed_profiles(fens, workers=2)` from a guarded Python script. No Spark or Java installation is required. This replaces the original independent, identical cell calculations with meaningful complete-position jobs.
