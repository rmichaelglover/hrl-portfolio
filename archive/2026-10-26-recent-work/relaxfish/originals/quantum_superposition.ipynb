```python
import numpy as np
from typing import Dict, List, Tuple

class QuantumTacticalSuperpositionEngine:
    """
    Models chess piece roles as a quantum-like wave function |Ψ⟩.
    Tracks state interference, calculates global field mixtures,
    and handles the collapse of role probabilities during tactical interaction.
    """
    def __init__(self, board_size: int = 8):
        self.N = board_size
        # The basis states of our tactical Hilbert space
        self.basis_roles = [
            "King-Shield", "Infiltrator", "Blockader",
            "Open-File-Anchor", "Diagonal-Lurker", "Escape-Choker"
        ]
        self.dim = len(self.basis_roles)

    def initialize_state_vector(self) -> np.ndarray:
        """
        Creates a clean state vector where roles exist in a uniform
        superposition—unobserved and fully fluid.
        """
        # A state vector for a single piece across all potential basis roles
        psi = np.ones(self.dim, dtype=complex)
        psi /= np.linalg.norm(psi) # Normalize so total probability equals 1.0
        return psi

    def apply_classical_field_interference(self, psi: np.ndarray, distance_to_king: float, is_blocked: bool) -> np.ndarray:
        """
        Applies environmental constraints as phase shifts and amplitude modifiers.
        The classical machine alters the quantum wave function shape.
        """
        modified_psi = psi.copy()

        # 1. Proximity to King alters the "King-Shield" state amplitude
        if distance_to_king <= 2.0:
            # Shift weight toward defensive basis states
            modified_psi[0] *= 1.8  # Amplify King-Shield potential
        else:
            modified_psi[0] *= 0.2  # Attenuate when far away

        # 2. Pinball-style blocking vectors introduce quantum phase shifts (interference)
        if is_blocked:
            # Complex phase shift represents tactical tension behind a friendly shield
            modified_psi[2] *= np.exp(1j * np.pi / 4) # Phase shift 'Blockader' role
            modified_psi[1] *= 0.1                    # Suppress 'Infiltrator' capability

        # Re-normalize the wave function to maintain unitary stability
        modified_psi /= np.linalg.norm(modified_psi)
        return modified_psi

    def compute_probability_distribution(self, psi: np.ndarray) -> np.ndarray:
        """
        Calculates the Born rule probabilities P(role) = |⟨role|Ψ⟩|^2
        to observe the likelihood of each strategic state.
        """
        # The absolute square of the complex amplitudes
        probabilities = np.abs(psi) ** 2
        return probabilities

    def collapse_tactical_state(self, probabilities: np.ndarray) -> str:
        """
        Forces the fluid superposition to collapse into a single, discrete
        classical role index when a forced tactical capture occurs.
        """
        chosen_idx = np.random.choice(self.dim, p=probabilities)
        return self.basis_roles[chosen_idx]

# ==========================================
# RUNTIME INTEGRATION ENGINE
# ==========================================
if __name__ == "__main__":
    # Simulate a single position on an 8x8 board
    engine = QuantumTacticalSuperpositionEngine()

    print("=== 1. Initializing Superposition ===")
    psi_initial = engine.initialize_state_vector()
    initial_probs = engine.compute_probability_distribution(psi_initial)
    for role, prob in zip(engine.basis_roles, initial_probs):
        print(f"  Role: {role:<18} | Probability: {prob:.4f}")

    print("\n=== 2. Opponent Plays Bongcloud (King Advances, Distance Closes) ===")
    # A piece close to an exposed, advancing King behind a pawn block
    psi_interfered = engine.apply_classical_field_interference(psi_initial, distance_to_king=1.41, is_blocked=True)
    interfered_probs = engine.compute_probability_distribution(psi_interfered)
    for role, prob in zip(engine.basis_roles, interfered_probs):
        print(f"  Role: {role:<18} | Probability: {prob:.4f}")

    print("\n=== 3. Tactical Conflict (The Wave Function Collapses) ===")
    for turn in range(1, 4):
        observed_role = engine.collapse_tactical_state(interfered_probs)
        print(f"  Turn {turn} Observation: Piece acts strictly as a '{observed_role}'")
```
