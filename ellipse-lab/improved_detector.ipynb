# Robust rotated ellipse fitting

A separate improved experiment; see REPORT.md for the 30-scene benchmark and failure cases. RANSAC performs global outlier rejection; local refinement follows. No HRL-specific advantage is claimed. Run beside study.py.


```python
import numpy as np
import matplotlib.pyplot as plt
from study import scene, original, robust, curve, metrics

points, truth = scene(seed=42)
baseline = original(points)
improved, ransac_fit = robust(points, seed=42)
for name, fit in [("Original grid", baseline), ("RANSAC", ransac_fit), ("Refined", improved)]:
    print(name, fit, metrics(fit, truth, points))
```

```python
fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(*points.T, s=12, c="gray", label="Unlabeled input")
for name, fit in [("Truth (evaluation only)", truth), ("Original", baseline), ("Refined", improved)]:
    xy = curve(fit)
    ax.plot(*xy.T, label=name)
ax.set(aspect="equal", xlim=(0,100), ylim=(0,100))
ax.legend()
plt.show()
```

## Further checks

Change scene seed, clutter count, sigma, and rotation. The detector is not guaranteed to recover the ellipse under heavy clutter. Use `python3 study.py` for the complete recorded comparison.
